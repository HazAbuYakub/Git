## <span style="color:yellow"> **GitHub kennenlernen** </span>

Nachdem wir verstanden haben, wie **Git lokal auf dem eigenen Rechner** funktioniert, kommt im nächsten Schritt **GitHub** dazu.

**GitHub** ist **keine Alternative** zu Git, sondern eine **Ergänzung**.

- **Git** verwaltet Versionen lokal auf dem Rechner
- **GitHub** speichert Git-Repositories online
- **GitHub** macht Zusammenarbeit und Datensicherung einfacher

Erst wenn die Grundlagen von Git sitzen, ergibt GitHub wirklich Sinn.

Wie sehen uns an:

1. Ein Repository auf GitHub erstellen  
2. Das lokale Repository mit GitHub verbinden  
3. Änderungen mit `push` hochladen  
4. Änderungen mit `pull` herunterladen

### <span style="color:plum"> **Was ist GitHub?** </span>

GitHub ist eine Plattform, auf der Git-Repositories online gespeichert werden können.

### Vorteile von GitHub
- Projekte online sichern
- Von mehreren Rechnern aus arbeiten
- Mit anderen Personen zusammenarbeiten
- Änderungen austauschen
- Projekte veröffentlichen

### Wichtig!
GitHub arbeitet **mit Git**.  
Ohne Git gibt es kein sinnvolles Arbeiten mit GitHub.

### <span style="color:plum">Lokales Git-Repository mit GitHub verbinden</span>

Wir haben bereits ein lokales Git-Repository erstellt und erste Commits gemacht.

Jetzt kommt GitHub dazu:

- **Schritt 1**
    - Auf GitHub ein neues Repository erstellen

- **Schritt 2**
  - Die URL des GitHub-Repositories kopieren

Beispiel:
    - `https://github.com/USERNAME/mein-projekt.git`


- **Schritt 3**
Das lokale Repository mit GitHub verbinden

Dazu verwenden wir:
  - `git remote add origin <URL>`


- **Schritt 4**
  - Die lokalen Commits zu GitHub hochladen

Dazu verwenden wir:
  - `git push -u origin main`


- **Schritt 5**
  - Spätere Änderungen abrufen

Dazu verwenden wir:
  - `git pull`

---

### <span style="color:plum">Code-Zelle oder Terminal-Zelle: Befehle</span>

```python
# Verbindet das lokale Repository mit dem GitHub-Repository
# "origin" ist der Standardname für das entfernte Repository
git remote add origin https://github.com/USERNAME/mein-projekt.git

# Lädt den aktuellen Branch zum ersten Mal auf GitHub hoch
# "-u" merkt sich die Verbindung zwischen lokalem und entferntem Branch
git push -u origin main

# Holt Änderungen vom GitHub-Repository auf den eigenen Rechner
git pull
```


Wenn wir lokal arbeiten, liegen unsere Änderungen zunächst nur auf unserem eigenen Rechner.

### `git push`
Mit `git push` senden wir unsere lokalen Commits zu GitHub.

Einfach gesagt:
> **push = hochladen**

### `git pull`
Mit `git pull` holen wir Änderungen von GitHub auf unseren Rechner.

Einfach gesagt:
> **pull = herunterladen**

### Merksatz
- **commit** = lokal speichern
- **push** = zu GitHub hochladen
- **pull** = von GitHub herunterladen

---

Bis jetzt haben wir direkt in unserem Projekt gearbeitet und Änderungen mit Commits gespeichert.

In der Praxis möchte man neue Ideen oder Funktionen oft erst ausprobieren, ohne den funktionierenden Stand zu verändern.

Genau dafür gibt es **Branches**.

### <span style="color:plum">Was ist ein Branch?</span>
Ein Branch ist ein separater Entwicklungszweig.

Man kann sich das wie eine Nebenstrecke vorstellen:
- Im Hauptzweig liegt der aktuelle Stand des Projekts
- In einem neuen Branch können wir Änderungen testen
- Wenn alles funktioniert, können wir die Änderungen zurück in den Hauptzweig übernehmen

### <span style="color:plum">Warum sind Branches nützlich?</span>
- Neues ausprobieren, ohne den Hauptstand zu gefährden
- Funktionen getrennt entwickeln
- Fehler beheben, ohne direkt im Hauptzweig zu arbeiten
- Sauber und sicher im Team zusammenarbeiten

### <span style="color:plum">Branch anzeigen und erstellen</span>

```python
#Zeigt an, auf welchem Branch wir uns gerade befinden
git branch

#Erstellt einen neuen Branch mit dem Namen "feature-login"
git branch feature-login


Mit `git branch` können wir uns bestehende Branches anzeigen lassen.
Mit `git branch feature-login` erstellen wir einen neuen Branch.

Wichtig:
Der Branch wird damit nur erstellt, aber wir arbeiten noch nicht automatisch darin.
```

### <span style="color:plum">In einen Branch wechseln</span>

```python
Nachdem ein Branch erstellt wurde, müssen wir in ihn wechseln.

Erst dann arbeiten wir wirklich in diesem neuen Zweig.

Ein neuer Branch ist zunächst nur angelegt.  
Mit `git switch` wechseln wir aktiv dorthin.

# Wechselt in den Branch "feature-login"
git switch feature-login
```

### <span style="color:plum">Branch erstellen und direkt wechseln</span>

```python
# Erstellt den Branch "feature-login" und wechselt direkt hinein
git switch -c feature-login
```

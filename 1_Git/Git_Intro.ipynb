### <span style="color:plum"> **Was ist Git?** </span>

Git ist ein Werkzeug zur **Versionsverwaltung** auf dem eigenen Rechner.  
Es speichert nicht einfach nur Dateien, sondern auch ihre Entwicklung.

- Änderungen an Dateien nachzuverfolgen,
- frühere Stände wiederherzustellen,
- sauber im Team zu arbeiten,
- und das typische Chaos mit Dateinamen wie `final`, `final_neu1` oder `wirklich_final3` zu vermeiden.


### <span style="color:plum"> **Was ist GitHub?** </span>

GitHub ist eine Plattform für die Zusammenarbeit an Softwareprojekten auf Basis von **Git**.

Dort können Teams und Einzelpersonen:
- Code speichern
- Änderungen nachverfolgen
- gemeinsam entwickeln
- Fehler, Aufgaben und Ideen dokumentieren


**Git** = Versionsverwaltung auf dem eigenen Rechner. <br>
**GitHub** = Webseite/Plattform, um Git-Repositories online zu speichern und gemeinsam zu nutzen.

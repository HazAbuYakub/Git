### <span style="color:plum"> **Was ist Git?** </span>

Git ist ein Werkzeug zur **Versionsverwaltung** auf dem eigenen Rechner.  
Es speichert nicht einfach nur Dateien, sondern auch ihre Entwicklung.

- Änderungen an Dateien nachzuverfolgen,

- sauber im Team zu arbeiten,
- und das typische Chaos mit Dateinamen wie `final`, `final_neu` oder `wirklich_final` zu vermeiden.


### <span style="color:plum"> **Was ist GitHub?** </span>

GitHub ist eine Plattform für die Zusammenarbeit an Softwareprojekten auf Basis von **Git**.

Dort können Teams und Einzelpersonen:

**Git** = Versionsverwaltung auf dem eigenen Rechner. <br>
**GitHub** = Webseite/Plattform, um Git-Repositories online zu speichern und gemeinsam zu nutzen.

## <span style="color:yellow"> **Kapitel 3: GitHub** </span>

### <span style="color:plum"> **Was ist GitHub?** </span>

GitHub ist eine Plattform, auf der Git-Repositories online gespeichert werden können.

### Vorteile von GitHub
- Projekte online sichern
- Von mehreren Rechnern aus arbeiten
- Mit anderen Personen zusammenarbeiten
- Änderungen austauschen
- Projekte veröffentlichen

### Wichtig!
GitHub arbeitet **mit Git**.  
Ohne Git gibt es kein sinnvolles Arbeiten mit GitHub.

### <span style="color:plum">Lokales Git-Repository mit GitHub verbinden</span>

Wir haben bereits ein lokales Git-Repository erstellt und erste Commits gemacht.

Jetzt kommt GitHub dazu:

- **Schritt 1**
    - Auf GitHub ein neues Repository erstellen

- **Schritt 2**
  - Die URL des GitHub-Repositories kopieren

Beispiel:
    - `https://github.com/USERNAME/mein-projekt.git`


- **Schritt 3**
Das lokale Repository mit GitHub verbinden

Dazu verwenden wir:
  - `git remote add origin <URL>`


- **Schritt 4**
  - Die lokalen Commits zu GitHub hochladen

Dazu verwenden wir:
  - `git push -u origin main`


- **Schritt 5**
  - Spätere Änderungen abrufen

Dazu verwenden wir:
  - `git pull`

---

### <span style="color:plum">Code-Zelle oder Terminal-Zelle: Befehle</span>

```python
# Verbindet das lokale Repository mit dem GitHub-Repository
# "origin" ist der Standardname für das entfernte Repository
git remote add origin https://github.com/USERNAME/mein-projekt.git

# Lädt den aktuellen Branch zum ersten Mal auf GitHub hoch
# "-u" merkt sich die Verbindung zwischen lokalem und entferntem Branch
git push -u origin main

# Holt Änderungen vom GitHub-Repository auf den eigenen Rechner
git pull
```


Wenn wir lokal arbeiten, liegen unsere Änderungen zunächst nur auf unserem eigenen Rechner.

### `git push`
Mit `git push` senden wir unsere lokalen Commits zu GitHub.

Einfach gesagt:
> **push = hochladen**

### `git pull`
Mit `git pull` holen wir Änderungen von GitHub auf unseren Rechner.

Einfach gesagt:
> **pull = herunterladen**

### Merksatz
- **commit** = lokal speichern
- **push** = zu GitHub hochladen
- **pull** = von GitHub herunterladen

---



Bis jetzt haben wir direkt in unserem Projekt gearbeitet und Änderungen mit Commits gespeichert.

In der Praxis möchte man neue Ideen oder Funktionen oft erst ausprobieren, ohne den funktionierenden Stand zu verändern.

Genau dafür gibt es **Branches**.

### Was ist ein Branch?
Ein Branch ist ein separater Entwicklungszweig.

Man kann sich das wie eine Nebenstrecke vorstellen:
- Im Hauptzweig liegt der aktuelle Stand des Projekts
- In einem neuen Branch können wir Änderungen testen
- Wenn alles funktioniert, können wir die Änderungen zurück in den Hauptzweig übernehmen

### Warum sind Branches nützlich?
- Neues ausprobieren, ohne den Hauptstand zu gefährden
- Funktionen getrennt entwickeln
- Fehler beheben, ohne direkt im Hauptzweig zu arbeiten
- Sauber und sicher im Team zusammenarbeiten

## <span style="color:yellow"> **Kapitel E: Datenschutz bei GitHub** </span>

### <span style="color:plum"> **Was man wissen sollte** </span>

Beim Datenschutz kommt es stark darauf an, **wie** GitHub genutzt wird.

Wichtige Punkte:

- **Öffentliche Repositories** sind öffentlich sichtbar. Code, Commits, Issues und oft auch Diskussionen können von anderen gelesen werden.
- **Private Repositories** sind nur für berechtigte Personen sichtbar, aber die Daten liegen trotzdem bei GitHub.
- GitHub beschreibt in seiner Datenschutzerklärung, wie personenbezogene Daten verarbeitet werden.
- Für Unternehmen, Schulen oder Behörden ist wichtig zu prüfen, **welche Daten hochgeladen werden dürfen**.
- Zugangsschutz wie **starke Passwörter** und **Zwei-Faktor-Authentifizierung** ist sehr wichtig.

Merksatz:  
**GitHub ist praktisch, aber man sollte keine sensiblen Daten unüberlegt hochladen.**

## <span style="color:yellow"> **Kapitel F: Darf ich Code aus Open-Source-Projekten nehmen?** </span>

### <span style="color:plum"> **Ja, aber nur mit Blick auf die Lizenz** </span>

Grundsätzlich gilt:

- **Ja**, du darfst Open-Source-Code oft verwenden
- **Aber**: Entscheidend ist immer die **Lizenz**

Wichtige Regeln:

- Wenn ein Projekt z. B. unter **MIT**, **Apache 2.0** oder **GPL** steht, darfst du den Code oft unter bestimmten Bedingungen nutzen.
- Diese Bedingungen können z. B. sein:
  - Autor nennen
  - Lizenztext beilegen
  - Änderungen kenntlich machen
  - bei manchen Lizenzen den eigenen Code ebenfalls offenlegen
- **Ohne klare Lizenz** solltest du nicht einfach annehmen, dass du den Code frei übernehmen darfst.
- Auf GitHub dürfen andere Nutzer öffentliche Repositories zwar ansehen und forken, aber das ist **nicht automatisch dasselbe wie freie Weiterverwendung in jedem Projekt**.

Merksatz:  
**Nicht nur schauen, ob der Code öffentlich ist – immer auch die Lizenz prüfen.**


---

### <span style="color:plum"> **Typische Situationen** </span>

### Beispiel 1: MIT-Lizenz
Du findest ein Projekt mit MIT-Lizenz.  
Dann darfst du den Code meist verwenden, verändern und weitergeben, wenn du die Lizenzhinweise beachtest.

### Beispiel 2: Kein Lizenzhinweis
Du findest ein öffentliches Repository ohne Lizenz.  
Dann solltest du den Code **nicht einfach kopieren und weiterverwenden**, nur weil er öffentlich sichtbar ist.

### Beispiel 3: Schule oder Firma
Du arbeitest an einem internen Projekt.  
Dann solltest du prüfen, ob vertrauliche Informationen, Zugangsdaten oder Kundendaten wirklich auf GitHub gespeichert werden dürfen.
